```python
import json

notebook_path = "Country-PCA_Project.ipynb"

with open(notebook_path, "r", encoding="utf-8") as f:
    notebook = json.load(f)

print("=" * 70)
print("📦 NOTEBOOK CELL SIZE ANALYSIS")
print("=" * 70)

for i, cell in enumerate(notebook["cells"], start=1):

    # کل اطلاعات ذخیره‌شده سلول
    cell_json = json.dumps(cell, ensure_ascii=False)

    total_bytes = len(cell_json.encode("utf-8"))
    total_kb = total_bytes / 1024
    total_mb = total_kb / 1024

    # فقط کد/Markdown
    source = "".join(cell.get("source", []))
    source_bytes = len(source.encode("utf-8"))
    source_kb = source_bytes / 1024

    # حجم خروجی‌ها
    outputs = cell.get("outputs", [])
    output_json = json.dumps(outputs, ensure_ascii=False)

    output_bytes = len(output_json.encode("utf-8"))
    output_kb = output_bytes / 1024

    print(
        f"Cell {i:3} | "
        f"Total: {total_kb:8.2f} KB | "
        f"Source: {source_kb:8.2f} KB | "
        f"Outputs: {output_kb:8.2f} KB"
    )

print("=" * 70)
```
